## **Section 1. Business Understanding**

### **1.1 Context**

Pasar mobil bekas di Arab Saudi diproyeksikan tumbuh dengan CAGR 4,5% berdasarkan nilai transaksi pada periode 2019–2025, dengan adopsi kanal penjualan online yang meningkat pesat (Ken Research, 2021). Proyeksi yang lebih baru menunjukkan tren ini terus berlanjut: pasar diperkirakan tumbuh dari USD 7,45 miliar pada 2026 menjadi USD 11,24 miliar pada 2031, dan kanal online menjadi kanal dengan pertumbuhan tercepat, dengan CAGR 13,61% (Mordor Intelligence, 2026).

Salah satu pemain online di pasar ini adalah Syarah.com, yang hingga 2021 menjalankan dua jalur penjualan sekaligus. Jalur pertama adalah iklan baris (classifieds), tempat dealer dan penjual perorangan memasang iklan dan menentukan harga sendiri. Jalur kedua adalah penjualan mobil bekas secara online langsung oleh Syarah, yang dimulai sejak 2019, di mana setiap unit dilengkapi laporan inspeksi, jaminan uang kembali lima hari, dan garansi satu tahun (Wamda, 2021).

Pada Agustus 2021, Syarah memperoleh pendanaan Series B sebesar USD 20 juta. Dana tersebut direncanakan untuk membangun pusat distribusi sendiri dan menambah stok hingga lebih dari 1.000 mobil bekas yang telah melalui inspeksi dan rekondisi (Wamda, 2021). Rencana ini menunjukkan arah Syarah untuk memperbesar porsi penjualan stok sendiri. Konsekuensinya, Syarah membutuhkan pasokan mobil bekas dalam jumlah besar, beserta cara yang konsisten untuk menaksir harga wajar setiap unit sebelum membelinya.

### **1.2 Company Profile**

Syarah.com adalah platform jual-beli mobil online asal Arab Saudi yang didirikan pada 2015 oleh Salah Sharef dan Fayez Alanazi, dan berkantor pusat di Riyadh. Syarah awalnya beroperasi sebagai situs iklan baris (classifieds) untuk mobil baru dan bekas. Pada 2018, platform ini mencatat lebih dari 1,7 juta kunjungan per bulan dan lebih dari 17 ribu iklan aktif, dengan lebih dari 400 dealer di seluruh Arab Saudi serta ribuan penjual perorangan yang memasang mobilnya. Selain menampilkan iklan, Syarah menyediakan layanan tambahan berupa laporan riwayat kendaraan Mojaz dan fasilitas pembiayaan mobil (Wamda, 2018).

Sejak 2019, Syarah memperluas model bisnisnya dengan menjual mobil bekas secara langsung melalui platform online-nya. Setiap mobil yang dijual dilengkapi laporan inspeksi gratis, jaminan uang kembali lima hari, dan garansi satu tahun. (Wamda, 2021). Dengan demikian, hingga 2021 Syarah berperan ganda: sebagai perantara antara penjual dan pembeli di jalur iklan baris, sekaligus sebagai penjual langsung mobil bekas.

### **1.3 Problem Statements**

Untuk mencapai target lebih dari 1.000 mobil bekas dalam stok, Syarah membutuhkan pasokan mobil dalam jumlah besar. Salah satu sumber pasokan yang paling dekat justru ada di platform Syarah sendiri, yaitu penjual di jalur iklan baris. Pada dataset ini, 2.596 dari 7.800 listing iklan baris (sekitar 33%) tidak mencantumkan harga dan berstatus “Negotiable” (على السوم, secara harfiah “berdasarkan tawaran”), yang berarti penjual tidak menetapkan harga awal dan menunggu calon pembeli mengajukan penawaran. Penjual seperti ini merupakan kandidat potensial untuk didekati Syarah sebagai sumber stok.

Masalahnya, listing Negotiable tidak memberikan acuan harga apa pun. Untuk mengajukan penawaran beli, Syarah harus menaksir harga wajar setiap unit secara manual, misalnya dengan membandingkan listing serupa satu per satu. Cara ini sulit diterapkan pada ribuan listing, hasilnya bergantung pada penilai sehingga tidak konsisten, dan berisiko menghasilkan harga yang meleset dari pasar. Penawaran yang terlalu tinggi menggerus margin Syarah saat mobil dijual kembali, sedangkan penawaran yang terlalu rendah membuat penjual menolak dan Syarah kehilangan pasokan.

Oleh karena itu, Syarah membutuhkan cara untuk menaksir harga pasar mobil bekas secara cepat, konsisten, dan berbasis data, dengan memanfaatkan atribut kendaraan yang sudah tersedia di setiap listing, seperti merek, tipe, tahun produksi, jarak tempuh, kapasitas mesin, dan kelengkapan fitur.

### **1.4 Goals**

Proyek ini bertujuan mengembangkan model machine learning regresi yang mampu menaksir harga pasar mobil bekas berdasarkan atribut kendaraan, seperti merek, tipe, tahun produksi, jarak tempuh, kapasitas mesin, dan kelengkapan fitur. Model ini dirancang untuk mendukung Syarah dalam memperbesar porsi penjualan stok sendiri melalui dua tahap pemanfaatan.

Pada tahap pertama, model diterapkan pada listing Negotiable yang sudah ada di jalur iklan baris. Karena atribut kendaraan pada listing ini sudah tersedia dari iklan yang dipasang penjual, model dapat langsung memberikan estimasi harga pasar untuk setiap unit tanpa perlu mengumpulkan data tambahan. Estimasi tersebut membantu tim Syarah memilih mobil yang layak dibeli dan menjadi titik awal penawaran kepada penjual, sehingga pengadaan stok dapat dilakukan lebih cepat dan konsisten tanpa penaksiran manual satu per satu.

Pada tahap kedua, yaitu setelah Syarah sepenuhnya beralih dari platform iklan baris menjadi penjual mobil bekas yang membeli stoknya langsung dari pemilik kendaraan, listing Negotiable tidak lagi tersedia sebagai sumber pasokan. Penjual akan datang langsung ke Syarah untuk menjual mobilnya. Pada tahap ini, model digunakan sebagai estimasi harga awal saat penjual mengisi data kendaraannya. Estimasi ini kemudian disesuaikan dengan hasil inspeksi kondisi mobil, serta margin dan biaya rekondisi, sebelum menjadi penawaran final kepada penjual.

Dengan demikian, model diharapkan dapat mempercepat pengadaan stok, menjaga konsistensi harga beli antar unit, dan mengurangi risiko penawaran yang terlalu tinggi maupun terlalu rendah dibanding harga pasar.

### **1.5 Analytical Approach**

Karena variabel target, Price, merupakan nilai numerik kontinu, permasalahan ini termasuk ke dalam kategori regresi. Model dikembangkan menggunakan data historis listing yang sudah memiliki harga, dengan mempelajari pola hubungan antara atribut kendaraan dan harganya. Pola ini kemudian diterapkan untuk menaksir harga pasar listing Negotiable, yang tidak memiliki harga.

Data latih berasal dari dua jalur penjualan Syarah. Sebagian besar adalah listing iklan baris berharga, dengan harga yang ditentukan oleh penjual (asking price). Sebagian kecil sisanya, sekitar 8% dari listing berharga, adalah listing mobil bekas yang dijual langsung oleh Syarah, dengan harga jual yang ditetapkan Syarah untuk mobil yang sudah diinspeksi dan bergaransi. Karena itu, Price pada dataset ini adalah harga yang ditawarkan di platform, bukan harga transaksi akhir maupun harga beli Syarah dari penjual. Model akan mempelajari pola harga pasar dari listing, dan estimasinya berfungsi sebagai acuan harga pasar yang masih perlu disesuaikan dengan hasil inspeksi, serta margin dan biaya rekondisi, sebelum menjadi penawaran beli.

Pendekatan ini mengasumsikan bahwa listing Negotiable memiliki karakteristik kendaraan yang sebanding dengan listing berharga. Asumsi ini akan divalidasi pada tahap EDA.

### **1.6 Metric Evaluation (Business Metric, Machine Learning Evaluation Metric)**

Karena permasalahan proyek ini adalah regresi, evaluasi model difokuskan pada seberapa dekat hasil prediksi dengan harga listing aktual. **Mean Absolute Percentage Error (MAPE) dan Mean Absolute Error (MAE)** digunakan bersamaan sebagai metrik utama karena saling melengkapi. MAPE menyatakan error sebagai persentase dari harga, sehingga performa model dapat dibandingkan secara adil antara mobil murah dan mobil mewah, mengingat rentang harga dalam data sangat lebar (median sekitar SAR 58.000 hingga lebih dari SAR 1 juta). Ukuran dalam persen ini juga selaras dengan cara Syarah menghitung margin dan potongan harga saat menyusun penawaran beli. MAE dapat langsung dibaca dalam satuan Riyal, sehingga memberi gambaran nyata seberapa jauh estimasi harga dapat meleset untuk setiap unit. Apabila kedua metrik memberikan hasil yang berbeda dalam membandingkan model, MAPE dijadikan dasar pemilihan model karena lebih adil terhadap rentang harga yang lebar.

Root Mean Squared Error (RMSE) dilaporkan sebagai metrik pendukung untuk memberi penalti lebih besar pada kesalahan prediksi yang ekstrem, mengingat distribusi Price yang right-skewed (skewness ≈ 3,7). R² turut dilaporkan untuk melihat proporsi variansi harga yang mampu dijelaskan model.

Seperti dijelaskan pada bagian Problem Statement, overestimation dan underestimation sama-sama merugikan Syarah dengan cara berbeda. Estimasi yang terlalu tinggi membuat Syarah membeli mobil di atas harga pasar sehingga margin saat dijual kembali tergerus. Estimasi yang terlalu rendah membuat penawaran ditolak penjual sehingga Syarah kehilangan pasokan stok. Oleh karena itu, selain besarnya error, arah error juga dipantau melalui rata-rata residual (aktual dikurangi prediksi) dan proporsi prediksi yang berada di atas maupun di bawah harga aktual.

### **1.7 Success Criteria**

**Business Metrics:** Model ditargetkan menghasilkan MAPE ≤ 20% pada data testing, yang menurut kategori Lewis (1982, dalam Chen et al., 2003) termasuk prediksi yang baik (good forecasting). Dengan tingkat error ini, estimasi model dinilai cukup dekat untuk dijadikan acuan awal penawaran beli, yang kemudian disesuaikan dengan hasil inspeksi kondisi mobil. Selain itu, model tidak boleh menunjukkan bias sistematis, yang ditandai dengan rata-rata residual mendekati nol serta proporsi overestimate dan underestimate yang relatif seimbang (masing-masing di kisaran 45–55%). Dengan begitu, estimasi tidak cenderung membuat Syarah membayar di atas harga pasar ataupun menawar terlalu rendah hingga ditolak penjual.

**Machine Learning Performance:** Sebagai acuan, Gollapalli et al. (2023), yang menggunakan dataset Syarah yang sama, melaporkan R² sebesar 0,74 untuk Linear Regression dan 0,82 untuk Random Forest sebagai model terbaik. Oleh karena itu, model akhir pada proyek ini ditargetkan mencapai R² ≥ 0,82 pada data testing, sekaligus menghasilkan MAPE dan MAE yang lebih rendah dibandingkan Linear Regression sebagai model pembanding, tanpa indikasi overfitting signifikan antara data training dan testing. Perbandingan dengan studi tersebut bersifat indikatif karena adanya perbedaan tahapan preprocessing dan skala target.
