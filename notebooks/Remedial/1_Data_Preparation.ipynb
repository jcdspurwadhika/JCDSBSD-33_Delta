## **Section 2. Data Cleaning**

### **2.1 Remove Irrelevant Columns**

```python
# Identifikasi jalur penjualan dari pola URL
channel = pd.Series(
    np.where(df['Link'].str.contains('/online/'), 'Penjualan Langsung', 'Iklan Baris'),
    name='Channel'
)

assert len(channel) == len(df)
assert channel.notna().all()

pd.crosstab(channel, df['Negotiable'], margins=True, margins_name='Total')
```

Sebelum kolom `Link` dihapus, pola URL dimanfaatkan untuk mengidentifikasi jalur penjualan setiap listing. Listing dengan URL *`syarah.com/online/...`* merupakan mobil bekas yang dijual langsung oleh Syarah, sedangkan sisanya merupakan iklan baris yang dipasang oleh pihak ketiga, yaitu dealer maupun penjual perorangan. Dari 8.248 listing, sebanyak **7.800 listing (94,6%)** berasal dari iklan baris dan **448 listing (5,4%)** berasal dari penjualan langsung oleh Syarah.

Seluruh **2.596 listing Negotiable** berasal dari iklan baris, atau sekitar 33% dari total iklan baris. Sebaliknya, tidak ada satu pun listing penjualan langsung Syarah yang berstatus Negotiable, karena harganya selalu ditetapkan oleh Syarah. Temuan ini menegaskan bahwa listing Negotiable merupakan mobil milik pihak ketiga yang belum memiliki harga, sehingga relevan sebagai kandidat sumber stok bagi Syarah.

Informasi jalur penjualan ini tidak digunakan sebagai fitur model, karena seluruh listing Negotiable yang menjadi sasaran prediksi berasal dari jalur yang sama, yaitu iklan baris.

```python
df.drop(["Link", "Condition"], axis=1, inplace=True)
df.head()
```

Kolom `Link` dan `Condition` dihapus karena tidak memberikan informasi yang relevan untuk memprediksi harga

- **`Link`**: berisi URL halaman iklan di syarah.com. Selain informasi jalur penjualan yang sudah diambil di atas, URL ini hanya berfungsi sebagai penanda unik tiap listing dan bukan atribut kendaraan, sehingga tidak digunakan sebagai fitur.

#### **2.3.2 Column Type**

Kolom `Type` berisi nama model kendaraan. Dataset ini merupakan hasil terjemahan dari versi bahasa Arab, sehingga sebagian nama model berpotensi keliru akibat proses konversi dari huruf Arab ke huruf Latin. Untuk memeriksanya, setiap nilai `Type` yang mencurigakan dicocokkan dengan nilai aslinya pada dataset versi bahasa Arab (`UsedCarsSA_Unclean_Ar.xlsx`), yang memiliki urutan baris yang sama dengan dataset versi Inggris.

```python
print(f"Jumlah Type unik: {df['Type'].nunique()}\n")

for make, types in df.groupby('Make')['Type'].unique().items():
    print(f"{make}: {sorted(types)}")
```

Dari daftar di atas, terdapat sejumlah nilai `Type` yang tidak lazim sebagai nama model kendaraan. Sebagian berupa kata bahasa Inggris yang tidak berkaitan dengan otomotif (misalnya `Furniture` dan `Gamble`), sebagian lagi berupa nama yang mirip dengan model asli namun penulisannya keliru (misalnya `Senta fe` dan `Kaptiva`). Selain itu, terdapat nilai yang mengandung karakter non-Latin dan nilai yang merupakan nama dealer.

```python
suspicious_types = [
    'Furniture', 'Senta fe', 'Ciocca', 'Copper', 'Vego', 'Carenz',
    'POS24', 'Gamble', 'Kaptiva', 'Cores', 'Abeka', 'Montero2',
    'Sylvian Bus', 'Capture', 'Osa Bus', 'Koranado', 'Montero1',
    '\u0650ACTIS V80', 'Behbehani',
]

# Cek tambahan: Type yang mengandung karakter non-Latin
print("Type dengan karakter non-ASCII:", df.loc[~df['Type'].map(str.isascii), 'Type'].unique())

anomali_type = df[df['Type'].isin(suspicious_types)]
print(f"Jumlah baris dengan Type mencurigakan: {len(anomali_type)}")
anomali_type.groupby(['Make', 'Type']).size().sort_values(ascending=False)
```

```python
# Dataset mentah versi Inggris dan Arab memiliki urutan baris yang sama
df_raw_en = pd.read_csv('../data/raw/UsedCarsSA_Unclean_EN.csv')
df_raw_ar = pd.read_excel('../data/raw/UsedCarsSA_Unclean_Ar.xlsx')

assert len(df_raw_en) == len(df_raw_ar)
assert (df_raw_en['Year'].values == df_raw_ar['Year'].values).all()

validasi = pd.DataFrame({
    'Make': df_raw_en['Make'],
    'Type_EN': df_raw_en['Type'],
    'Type_AR': df_raw_ar['Type'],
})
validasi = validasi[validasi['Type_EN'].isin(suspicious_types)].drop_duplicates()
validasi.sort_values('Make').reset_index(drop=True)
```

```python
# Mapping Type yang keliru ke nama model yang benar
# Divalidasi dengan dataset versi Arab dan halaman kategori syarah.com
type_mapping = {
    # Kesalahan konversi nama Arab ke Inggris
    'Furniture': 'Fortuner',         # فورتشنر - Toyota
    'Senta fe': 'Santa Fe',          # سنتافي - Hyundai
    'Ciocca': 'Sequoia',             # سيكويا - Toyota
    'Copper': 'Cooper',              # كوبر - MINI
    'Vego': 'Figo',                  # فيجو - Ford
    'Carenz': 'Carens',              # كارينز - Kia
    'POS24': 'Short Bonnet',         # بوز 24 - Mercedes truk klasik (1959-1995)
    'Gamble': 'Ghibli',              # جيبلي - Maserati
    'Kaptiva': 'Captiva',            # كابتيفا - Chevrolet
    'Cores': 'Quoris',               # كوريس - Kia
    'Abeka': 'Epica',                # ابيكا - Chevrolet
    'Montero2': 'Mountaineer',       # ماونتينير - Mercury
    'Sylvian Bus': 'Civilian',       # باص سيلفيان - Nissan
    'Capture': 'Captur',             # كابتشر - Renault
    'Osa Bus': 'Rosa',               # باص روزا - Mitsubishi
    'Koranado': 'Korando',           # كوراندو - SsangYong
    'Montero1': 'Montego',           # مونتيغو - Mercury

    # Karakter Arab tersembunyi (kasrah, U+0650) di depan nama
    '\u0650ACTIS V80': 'ACTIS V80',  # FAW
}

# Nilai yang bukan nama model -> di-drop
drop_types = ['Behbehani']           # بهبهاني - nama dealer

# Pastikan semua Type mencurigakan sudah punya penanganan
assert set(type_mapping) | set(drop_types) == set(suspicious_types)

# Perbaikan
before = len(df)
df = df[~df['Type'].isin(drop_types)].reset_index(drop=True)
df['Type'] = df['Type'].replace(type_mapping)

# Validasi
assert not df['Type'].isin(suspicious_types).any(), "Masih ada Type mencurigakan"
assert df['Type'].map(str.isascii).all(), "Masih ada karakter non-ASCII di Type"
print(f"{before} -> {len(df)} baris")
```

Ditemukan **296 baris** dengan nilai `Type` yang tidak sesuai dengan nama model mobil yang sebenarnya. Hampir seluruhnya berasal dari kesalahan konversi nama model dari huruf Arab ke huruf Latin. Kesalahan ini terjadi karena huruf Arab tidak membedakan beberapa bunyi, misalnya ب dapat terbaca sebagai B atau P, dan ف sebagai F atau V, sehingga nama yang dihasilkan meleset dari nama aslinya (contoh: `Vego` untuk Ford Figo). Pemeriksaan pertama difokuskan pada listing dengan harga sangat rendah, dengan membandingkan listing di bawah 5.000 SAR dan listing pada rentang 5.000–10.000 SAR.

```python
priced = df[df['Price'].notna()]

print(priced['Price'].describe().round(0))
print()
print(f"Price < 5,000          : {(priced['Price'] < 5_000).sum()} listing")
print(f"5,000 <= Price < 10,000: {priced['Price'].between(5_000, 10_000, inclusive='left').sum()} listing")
```

Harga minimum pada data adalah 0 SAR, padahal kuartil bawah (25%) sudah berada di 35.000 SAR. Terdapat **96 listing** dengan harga di bawah 5.000 SAR dan 58 listing pada rentang 5.000–10.000 SAR. 

Untuk mengetahui apakah harga-harga rendah ini wajar, listing di bawah 5.000 SAR dibandingkan dengan harga normal dari model dan tahun yang sama.

```python
low = priced[priced['Price'] < 5_000]

# Median dan jumlah listing normal (>= 5.000) untuk Make, Type, dan Year yang sama
normal_ref = (priced[priced['Price'] >= 5_000]
              .groupby(['Make', 'Type', 'Year'])['Price']
              .agg(Median_Normal='median', N_Normal='size'))

compare = low.merge(normal_ref, left_on=['Make', 'Type', 'Year'], right_index=True, how='left')
compare['Ratio'] = (compare['Price'] / compare['Median_Normal']).round(3)

# Hanya gunakan pembanding dengan minimal 3 listing agar median lebih stabil
reliable = compare[compare['N_Normal'] >= 3]
print(f"Listing dengan pembanding >= 3: {len(reliable)} dari {len(low)}")
print(f"Median rasio harga terhadap harga normal: {reliable['Ratio'].median():.1%}")

(reliable[['Make', 'Type', 'Year', 'Mileage', 'Price', 'Median_Normal', 'N_Normal', 'Ratio']]
    .sort_values('Price')
    .head(15))
```

```python
print(f"Median tahun listing < 5.000 SAR: {low['Year'].median():.0f}")
print(f"Listing tahun 2018 ke atas: {(low['Year'] >= 2018).sum()} dari {len(low)}")
```

Listing dengan harga di bawah 5.000 SAR didominasi oleh mobil keluaran baru, dengan median tahun 2019 dan 77 dari 96 listing merupakan mobil tahun 2018 ke atas. Untuk memeriksa kewajarannya, harga setiap listing dibandingkan dengan median harga listing normal (di atas 5.000 SAR) dari merek, model, dan tahun yang sama. Agar median pembanding cukup stabil, perbandingan hanya dilakukan pada listing yang memiliki minimal 3 listing pembanding, yaitu sebanyak 54 listing.

Hasilnya, harga listing tersebut umumnya hanya sekitar 2% dari harga normal. Contohnya, Hyundai Elantra 2019 tercatat seharga 993 SAR, sedangkan median harga normal untuk model dan tahun yang sama adalah 50.250 SAR (dari 36 listing pembanding). Selisih sebesar ini tidak mungkin terjadi pada transaksi jual beli mobil, sehingga nilai tersebut tidak mencerminkan harga jual yang sebenarnya. Penyebab pastinya tidak dapat dipastikan dari data, namun kemungkinan berasal dari kesalahan input oleh penjual, misalnya mengisi nilai asal seperti 0 atau 1 pada kolom harga, atau salah ketik jumlah digit.

Sebaliknya, listing pada rentang 5.000–10.000 SAR didominasi oleh mobil tua dengan jarak tempuh tinggi, dengan median tahun sekitar 2005 dan median jarak tempuh sekitar 217.000 km. Dengan demikian, batas 5.000 SAR dapat digunakan untuk memisahkan harga yang tidak wajar dari harga yang wajar.

```python
before = len(df)

# Drop listing dengan harga tidak wajar (< 5.000 SAR)
# Listing Negotiable (Price NaN) tetap dipertahankan
df = df[~(df['Price'] < 5_000)].reset_index(drop=True)

# Validasi
assert (df['Price'].dropna() >= 5_000).all(), "Masih ada Price < 5.000"
assert df['Price'].isna().sum() > 0, "Listing Negotiable ikut terhapus"
print(f"{before} -> {len(df)} baris ({before - len(df)} listing dihapus)")
```

Sebanyak **96 listing** dengan harga di bawah 5.000 SAR dihapus dari dataset. Karena jumlahnya sangat sedikit dan dampaknya terhadap model kecil, listing-listing ini tidak dihapus dan dicatat sebagai keterbatasan data. Penanganan outlier yang lebih sistematis dapat dilakukan pada tahap modeling setelah train-test split.

#### **2.5.2 Column Mileage**

Kolom `Mileage` diperiksa menggunakan batas berdasarkan logika domain. Sebagai acuan, panduan dari Syarah menyebutkan bahwa jarak tempuh normal mobil di Arab Saudi berkisar 15.000–20.000 km per tahun, sedangkan jarak tempuh di bawah 1.000 km per tahun menandakan mobil jarang digunakan [1]. Pemeriksaan dilakukan pada dua sisi, yaitu jarak tempuh yang terlalu tinggi dan jarak tempuh yang terlalu rendah untuk umur mobilnya.

[1] Syarah, *Natural Mileage per Year*, https://syarah.com/carsguide/natural-mileage-per-year/

```python
print(df['Mileage'].describe().round(0))
```

Sebagian besar listing memiliki jarak tempuh yang wajar, dengan median sekitar 103.000 km dan 75% data berada di bawah 197.000 km. Namun, nilai maksimumnya mencapai 20.000.000 km, yang menunjukkan adanya nilai ekstrem di sisi atas.

```python
high_mileage = df[df['Mileage'] > 1_000_000]
print(f"Mileage > 1,000,000 km: {len(high_mileage)} listing")

high_mileage = high_mileage.assign(
    Km_per_Year=(high_mileage['Mileage'] / (2021 - high_mileage['Year']).clip(lower=1)).round(0)
)
high_mileage[['Make', 'Type', 'Year', 'Mileage', 'Km_per_Year']].sort_values('Year')
```

Terdapat **44 listing** dengan jarak tempuh di atas 1.000.000 km, setara dengan sekitar 50 tahun pemakaian normal. Sebagian besar merupakan mobil keluaran 2005–2018 (median tahun 2008), sehingga rata-rata jarak tempuhnya mencapai ratusan ribu km per tahun, jauh di atas pemakaian normal. Contohnya, Hyundai Azera 2018 tercatat 1.100.000 km, atau sekitar 367.000 km per tahun. Banyak di antaranya juga berupa angka bulat atau berpola, seperti 3.000.000, 4.000.000, 1.111.111, dan 9.999.999, yang lebih menyerupai isian asal dibandingkan hasil pembacaan odometer. Satu-satunya pengecualian yang masih mungkin wajar adalah Toyota Dyna 1987 (1.100.000 km), sebuah truk berumur 34 tahun dengan rata-rata sekitar 32.000 km per tahun, namun dampaknya kecil karena hanya satu listing.

```python
old_low = df[(2021 - df['Year'] >= 5) & (df['Mileage'] < 1_000)]
print(f"Mobil >= 5 tahun dengan Mileage < 1,000 km: {len(old_low)} listing")
old_low['Mileage'].value_counts().head(10)
```

Di sisi lain, terdapat **798 listing** mobil berumur 5 tahun atau lebih dengan jarak tempuh di bawah 1.000 km, atau kurang dari 200 km per tahun. Angka ini jauh di bawah ambang "jarang digunakan" menurut panduan Syarah. Nilai yang paling sering muncul adalah angka bulat kecil seperti 300, 100, 400, dan 200, yang mengindikasikan penjual kemungkinan menuliskan jarak tempuh dalam satuan ribu km (misalnya 300 untuk 300.000 km). Namun, hal ini tidak dapat dipastikan, karena sebagian kecil mobil memang mungkin jarang digunakan.

```python
before_nan = df['Mileage'].isna().sum()

invalid_mileage = (
    (df['Mileage'] > 1_000_000) |
    ((2021 - df['Year'] >= 5) & (df['Mileage'] < 1_000))
)
df.loc[invalid_mileage, 'Mileage'] = np.nan

# Validasi
assert (df['Mileage'].dropna() <= 1_000_000).all()
assert not ((2021 - df['Year'] >= 5) & (df['Mileage'] < 1_000)).any()
print(f"Mileage diubah menjadi NaN: {df['Mileage'].isna().sum() - before_nan} listing")
print(f"Jumlah baris tetap: {len(df)}")
```

Nilai `Mileage` yang tidak wajar pada kedua kelompok tersebut (total 842 listing) **diubah menjadi NaN**, bukan dibiarkan, dihapus barisnya, atau dikoreksi secara manual. Jika dibiarkan, model akan mempelajari angka yang salah, misalnya menganggap mobil dengan jarak tempuh 3.500.000 km sebagai mobil yang sangat sering dipakai, padahal angka tersebut hanya kesalahan input. Menghapus barisnya akan menghilangkan sekitar 10% data, padahal atribut lain pada baris-baris tersebut, termasuk merek, model, tahun, dan harga, masih valid dan berguna bagi model. Sementara itu, mengoreksi nilainya secara manual (misalnya mengalikan 300 menjadi 300.000 km) hanya berdasarkan dugaan yang tidak dapat dipastikan, sehingga berisiko memasukkan data yang keliru.

Dengan mengubahnya menjadi NaN, nilai yang tidak dapat dipercaya dibuang tanpa kehilangan informasi lain pada baris yang sama. Nilai kosong ini akan diimputasi pada tahap modeling setelah train-test split, bersama nilai kosong pada kolom lain yang telah diidentifikasi di bagian 2.4, misalnya menggunakan median `Mileage` dari mobil dengan tahun yang sama pada data train, karena jarak tempuh sangat bergantung pada umur mobil. Pendekatan ini juga memudahkan penanganan data baru saat model digunakan. Jika suatu saat terdapat listing baru dengan jarak tempuh tidak wajar, misalnya di atas 1.000.000 km, aturan yang sama dapat langsung diterapkan: nilainya diubah menjadi NaN lalu diimputasi oleh pipeline. Sebaliknya, jika baris tersebut dihapus saat training, perlu ada aturan terpisah untuk menangani input seperti itu, misalnya dengan menolaknya, karena model tidak pernah mempelajari kasus serupa.

Batas yang digunakan (di atas 1.000.000 km, serta di bawah 1.000 km untuk mobil berumur 5 tahun atau lebih) ditentukan berdasarkan logika domain, bukan dihitung dari distribusi data. Oleh karena itu, penanganan ini aman dilakukan sebelum train-test split tanpa menimbulkan data leakage.
